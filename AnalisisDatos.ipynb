```python
import pandas as pd
from IPython.display import display

archivo_ventas = "ventas_libreria_cultural - ventas_libreria_cultural.csv"
df_libros = pd.read_csv(archivo_ventas, index_col=False)

print(f"Dimensiones: {df_libros.shape}")
display(df_libros.head())
```

## Selección, acceso y duplicados

Se conserva `df_libros` como copia original para consultar la fila con índice 310. Después se eliminan las filas repetidas y las exploraciones siguientes trabajan con `df_limpio`.

```python
# Selección unidimensional y multivariable
serie_generos = df_libros["Genero_Libro"]
tabla_sucursal_formato_precio = df_libros[["Sucursal", "Formato", "Precio_Lista"]]

print("Serie Genero_Libro:")
display(serie_generos.head())
print("Tabla con tres variables:")
display(tabla_sucursal_formato_precio.head())

# Acceso por etiqueta: conserva todas las columnas de la fila con índice 310
transaccion_indice_310 = df_libros.loc[310]
display(transaccion_indice_310)

# Eliminación de filas completamente duplicadas
df_limpio = df_libros.drop_duplicates().copy()
print(f"Filas originales: {len(df_libros)}")
print(f"Filas después de drop_duplicates(): {len(df_limpio)}")
print(f"Filas duplicadas eliminadas: {len(df_libros) - len(df_limpio)}")
```

## Filtrado de ventas

Cada consulta parte de `df_limpio`. Se guardan los resultados completos en variables y se muestran sus dimensiones y primeras filas para mantener legible el cuaderno.

```python
# Compras mayoristas: más de 3 unidades
compras_mayoristas = df_limpio[df_limpio["Unidades_Vendidas"] > 3]

# Operaciones de Campus Universitario y Galería de Arte
ventas_sucursales_seleccionadas = df_limpio[
    df_limpio["Sucursal"].isin(["Campus Universitario", "Galería de Arte"])
]

# Precios entre $45.000 y $75.000, inclusive
ventas_rango_precios = df_limpio[
    df_limpio["Precio_Lista"].between(45000, 75000, inclusive="both")
]

# Transferencia QR con al menos 2 unidades
ventas_qr_multiples = df_limpio.query(
    "Metodo_Pago == 'Transferencia QR' and Unidades_Vendidas >= 2"
)

# Libros de género Ensayo con precio superior a $50.000
ventas_ensayo_premium = df_limpio[
    df_limpio["Genero_Libro"].str.startswith("Ensayo")
    & (df_limpio["Precio_Lista"] > 50000)
]

for nombre, resultado in [
    ("Compras mayoristas", compras_mayoristas),
    ("Sucursales seleccionadas", ventas_sucursales_seleccionadas),
    ("Rango de precios", ventas_rango_precios),
    ("Transferencia QR", ventas_qr_multiples),
    ("Ensayo sobre $50.000", ventas_ensayo_premium),
]:
    print(f"{nombre}: {len(resultado)} filas")
    display(resultado.head())
```

## Imputación y preparación de variables

El precio faltante se reemplaza por la media de los precios disponibles y el descuento faltante por `0.0`. Luego se calcula el total neto por transacción, se renombran las columnas solicitadas y se reorganiza el DataFrame.

```python
# Imputación de valores faltantes sobre la base sin duplicados
promedio_precio = df_limpio["Precio_Lista"].mean()
df_limpio["Precio_Lista"] = df_limpio["Precio_Lista"].fillna(promedio_precio)
df_limpio["Descuento_Aplicado"] = df_limpio["Descuento_Aplicado"].fillna(0.0)

# Subtotal neto por fila
def calcular_subtotal_neto(fila):
    return (
        fila["Precio_Lista"]
        * (1 - fila["Descuento_Aplicado"])
        * fila["Unidades_Vendidas"]
    )


df_limpio["Total_Venta"] = df_limpio.apply(calcular_subtotal_neto, axis=1)

# Renombrar variables y colocar ID_Transaccion y Total_Venta al inicio
df_libros_final = df_limpio.rename(
    columns={"Genero_Libro": "Categoria", "Descuento_Aplicado": "Descuento"}
).copy()
columnas_iniciales = ["ID_Transaccion", "Total_Venta"]
columnas_restantes = [
    columna for columna in df_libros_final.columns if columna not in columnas_iniciales
]
df_libros_final = df_libros_final[columnas_iniciales + columnas_restantes]

print(f"Promedio usado para Precio_Lista: ${promedio_precio:,.2f}")
print("Nulos después de imputar:")
display(df_libros_final[["Precio_Lista", "Descuento"]].isna().sum())
display(df_libros_final.head())
```

## Agrupaciones y métricas

Los resúmenes se calculan sobre `df_libros_final`, después de limpiar, imputar y crear `Total_Venta`.

```python
# Promedio del precio y facturación total por sucursal
facturacion_por_sucursal = df_libros_final.groupby("Sucursal").agg(
    Precio_Promedio=("Precio_Lista", "mean"),
    Facturacion_Total=("Total_Venta", "sum"),
)
display(facturacion_por_sucursal)

# Rango (máximo - mínimo) de precio y venta total por formato
def rango(series):
    return series.max() - series.min()


rango_por_formato = df_libros_final.groupby("Formato").agg(
    {"Precio_Lista": rango, "Total_Venta": rango}
)
display(rango_por_formato)

# Promedio de Total_Venta y unidades totales por sucursal y categoría
resumen_sucursal_categoria = df_libros_final.groupby(
    ["Sucursal", "Categoria"]
).agg(
    Promedio_Total_Venta=("Total_Venta", "mean"),
    Unidades_Totales=("Unidades_Vendidas", "sum"),
)
display(resumen_sucursal_categoria)
```
